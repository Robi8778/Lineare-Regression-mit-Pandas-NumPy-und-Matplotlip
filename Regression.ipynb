```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = pd.read_csv('linreg_data.csv')

X = data['X'].values
Y = data['Y'].values

def linear_regression(x, y):
    n = len(x)
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    
    numerator = np.sum((x - x_mean) * (y - y_mean))
    denominator = np.sum((x - x_mean) ** 2)
    m = numerator / denominator
    
    b = y_mean - m * x_mean
    
    return m, b

m, b = linear_regression(X, Y)

Y_line = m * X + b

plt.scatter(X, Y, color='blue', label='Datenpunkte')
plt.plot(X, Y_line, color='red', label=f'Regressionsgerade: y = {m:.4f}x + {b:.4f}')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()

print(f"Steigung m: {m:.4f}")
print(f"Achsenabschnitt b: {b:.4f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress


data = pd.read_csv('linreg_data.csv')


X = data["X"]
Y = data["Y"]


slope, intercept, r_value, p_value, std_err = linregress(X, Y)

print("Steigung", slope)
print("Achsenabschnitt", intercept)
print("Korrelationskoeffizient r:", r_value)
print("p-Wert:", p_value)
print("Standardfehler:", std_err)


plt.scatter(X, Y, label="Datenpunkte")
x_sorted = sorted(X)
y_fit = [intercept + slope * x for x in x_sorted]
plt.plot(x_sorted, y_fit, color='red', label="Regressionsgerade", linewidth=2)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()


# Aufgabe 3: Vorhersage bei X = 2

x_new = 2
y_pred = intercept + slope * x_new
print(f"Vorhersage für X=2: Y ≈ {y_pred:.4f}")


# Aufgabe 4: Glättung der Daten

data["Y_smooth"] = intercept + slope * data["X"]

print("\n geglättete Werte:")
print(data[["X", "Y", "Y_smooth"]].head())


```
